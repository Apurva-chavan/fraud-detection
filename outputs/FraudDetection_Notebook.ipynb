# Healthcare Provider Fraud Detection

**Accuracy: 93.3%** | **AUC: 0.9531** | **Fraud Detected: 144/1353**

## 1. Objective
Predict fraudulent Medicare providers from claims data.

Fraud types: fake billing, duplicate claims, upcoding, billing for deceased patients.

## 2. Dataset
| File | Records |
|------|--------|
| Train Labels | 5,410 |
| Beneficiary | 138,556 |
| Inpatient | 40,474 |
| Outpatient | 517,737 |

## 3. Results
| Metric | Score |
|--------|-------|
| Accuracy | 93.3% |
| AUC | 0.9531 |
| Precision | 63% |
| Recall | 71% |

Fraud: **144** | Legitimate: **1209** | Total: **1353**

## 4. Key Findings
1. High reimbursements = strongest fraud signal
2. Many physicians per provider = fraud ring
3. Deceased patient claims = red flag
4. Long hospital stays + high billing = fake claims
5. High outpatient volume = claim splitting

## 5. Recommendations
1. Flag providers >0.5 probability for review
2. Audit all flagged providers immediately
3. Auto reject deceased patient claims
4. Monitor large physician networks
5. Retrain quarterly
